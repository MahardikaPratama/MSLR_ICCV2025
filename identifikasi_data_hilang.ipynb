# Identifikasi Data Train/Dev yang Hilang
Notebook ini akan membandingkan daftar video yang seharusnya ada (berdasarkan metadata list CSV asli yang sudah di-*preprocess* menjadi JSON) dengan data *skeleton* aktual yang berhasil diekstrak dan tersimpan di dalam file `.pkl`.

```python
import json
import pickle

# 1. Load data metadata (JSON)
try:
    with open('./datasets/mslr2025/sd_train_info.json', 'r') as f:
        train_info = json.load(f)
    
    with open('./datasets/mslr2025/sd_dev_info.json', 'r') as f:
        dev_info = json.load(f)
except FileNotFoundError:
    print("Error: File info JSON tidak ditemukan. Pastikan Anda sudah menjalankan mslr_process.py!")
    train_info, dev_info = [], []

# 2. Load data skeleton aktual (Pickle)
try:
    with open('./datasets/pose_bisindo_train_dev.pkl', 'rb') as f:
        kps_global = pickle.load(f)
    available_video_ids = set(kps_global.keys())
except FileNotFoundError:
    print("Error: File pose_bisindo_train_dev.pkl tidak ditemukan di folder ./datasets/!")
    available_video_ids = set()

# 3. Cari yang hilang di set Train
missing_train = []
for item in train_info:
    if item['video_id'] not in available_video_ids:
        missing_train.append(item)

# 4. Cari yang hilang di set Dev
missing_dev = []
for item in dev_info:
    if item['video_id'] not in available_video_ids:
        missing_dev.append(item)

# --- HASIL ---
print(f"Total data Train di metadata: {len(train_info)}")
print(f"Total data Train aktual yang ter-load: {len(train_info) - len(missing_train)}")
print(f"Data Train yang gagal di-load / hilang: {len(missing_train)}\n")
if missing_train:
    print("Daftar Train yang hilang:")
    for m in missing_train:
        print(f" - ID Video: {m['video_id']} | Gloss: {m['text']}")

print("\n" + "="*50 + "\n")

print(f"Total data Dev di metadata: {len(dev_info)}")
print(f"Total data Dev aktual yang ter-load: {len(dev_info) - len(missing_dev)}")
print(f"Data Dev yang gagal di-load / hilang: {len(missing_dev)}\n")
if missing_dev:
    print("Daftar Dev yang hilang:")
    for m in missing_dev:
        print(f" - ID Video: {m['video_id']} | Gloss: {m['text']}")
```
